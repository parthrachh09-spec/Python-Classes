```python
phone_numbers = {}
n = int(input('How many phone numbers do you have? '))
for i in range(n):
    number = input('Enter phone numbers:')
    if number not in phone_numbers:
        phone_numbers[number] = 1
    else:
        phone_numbers[number] = phone_numbers[number] + 1
print('Unique Phone Numbers:')
for number in phone_numbers:
    print(number)
print('Total uniques numbers:', len(phone_numbers))
print('Duplicate Numbers:')

for number in phone_numbers:
    if phone_numbers[number] > 1:
        print(number, 'was entered' , phone_numbers[number], 'times' )

```
