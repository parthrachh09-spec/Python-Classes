```python
def change(paid, price):
    return paid - price
price = 25
money = 0
print("==== VENDING MACHINE ====")
print("Snack price:",price)
print("Insert 11, 5, 10 or 25")

while money < price:
    coin = int(input("Insert coin:"))

    if coin == 1 or coin == 5 or coin == 10 or coin == 25:
        money =  money = coin
        print("Money inserted:", money)
    else:
        print("Invalid coin!")

print("/nSnack is dispensed!")
if money > price:
    print("Change returned to user:", change(money, price))

print("Thank you!")
```
