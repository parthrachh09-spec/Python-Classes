```python
questions = (
    'What is the capital of India?',
    'How many days are there in a week?',
    'What is the loudest animal in the Earth?',
    'What is 2 times 0?',
    'Which language are we learning?'
)
answers = (
    'Dehli',
    '7',
    'A Sperm Whale',
    '0',
    'python'
)
score = 0
print('Welcome to the Quiz!')
print("---------------------")


for i in range(5):
    print(questions[i])
    user_answer = input('Enter your answer')

    if user_answer== answers[i]:
        print('Correct')
        score = score + 1

    else:
        print('Wrong!')
        print('Correct answer is:'< answers[i])

    print()
print('Quiz Completed!')
print('Your Score:',score, 'out of 5')
```
