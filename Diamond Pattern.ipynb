```python
n = int(input("Enter number of rows: "))

number = 1
for i in range(1, n + 1):
    print(" " * (n-i), end="")

    for j in range(i):
        print(number, end=" ")
        number = number + 1

    print()


for i in range(n - 1, 0, -1):
    print(" " * (n-i), end="")

    for j in range (i):
        print(number , end=" ")
        number = number + 1

    print()
    
```
